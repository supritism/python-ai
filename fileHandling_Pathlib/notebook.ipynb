## PART 1 — File Handling
File handling matters in AI Engineering because every evaluation pipeline reads and writes files constantly like :
- loading eval datasets from jsonl files
- saving results after each run
- reading Config files
- Organising output folders by model and run

### Opening Files — the 'with' statement

```
with open("file.txt", "r") as f:
content = f.read()
```

- `with` automatically closes the file when the block ends - even if any error occurs. 

### File Modes
- open("file.txt", "r")    # read only — default
- open("file.txt", "w")    # write — creates file, overwrites if exists
- open("file.txt", "a")    # append — adds to end, creates if missing
- open("file.txt", "x")    # exclusive create — fails if file already exists
- open("file.txt", "rb")   # read binary — for images, PDFs, non-text files
- open("file.txt", "wb")   # write binary

The ones we use most in AI work is 'r', 'w' and 'a'

### Reading Files - Text mode
1. Read/Load entire file as ONE string into memory - f.read() : (Will cause memory crash for large files)
    with open("textdetails.txt", "r") as f:
        content = f.read()

2. Read/Load entire file as ONE List into memory - f.readlines() : (Will cause memory crash for large files)
    with open("eval_cases.jsonl", "r") as f:
        lines = f.readlines()    # ["line1\n", "line2\n", "line3\n"]

3. Read/Load one line at a time, discard it and load next : (Always use this for large files/JSONL datasets)
    with open("eval_cases.jsonl", "r") as f:
        for line in f :
        line_content = line.strip()

### Writing Files - Text Mode
1. Write a String
    with open("output.txt", "w") as f:
        f.write("Evaluation run complete\n")

2. Write multiple lines
lines = ["Case1 = Passed\n" , "Case2 = Failed\n"]
    with open("output.txt", "w") as f:
        f.writelines(lines)

3. Append to existing File
    with open("output.txt", "a") as f:
        f.write("Case3 = Failed\n")


### Reading and Writing : Brinary Mode (non text files like Images, PDFs etc)
 1. Read : 
    with open("output.bin", "rb") as f:
         content = f.read()

2. Write : 
    with open("output.bin", "wb") as f:
        f.write(data)

Note - Binary mode = raw bytes, no encoding/decoding, no line ending conversion. Add 'b' to any file mode — "rb", "wb", "ab"

````python
# Real AI Eval Pipeline — Full File Organization Pattern - V.V Imp
    # This is how production eval pipelines organize their files

import json
from pathlib import Path
from datetime import date

def setup_eval_run(base_dir: str, run_name: str) -> dict:

    # build folder structure
    base        = Path(base_dir)
    run_dir     = base / run_name / str(date.today())
    input_dir   = run_dir / "inputs"
    output_dir  = run_dir / "outputs"
    log_dir     = run_dir / "logs"

    # create all folders
    for folder in [input_dir, output_dir, log_dir]:
        folder.mkdir(parents=True, exist_ok=True)

    print(f"Eval run setup at: {run_dir}")
    return {
        "run_dir":    run_dir,
        "input_dir":  input_dir,
        "output_dir": output_dir,
        "log_dir":    log_dir
    }

def run_eval_pipeline(dataset_path: str, base_dir: str):

    # setup folders
    paths = setup_eval_run(base_dir, "eval_run")

    # check dataset exists
    dataset = Path(dataset_path)
    if not dataset.exists():
        print(f"Dataset not found: {dataset_path}")
        return

    # load all cases
    with open(dataset, "r") as f:
        cases = [json.loads(line.strip()) for line in f]
    print(f"Loaded {len(cases)} cases from {dataset.name}")

    # process and save results
    output_file = paths["output_dir"] / f"{dataset.stem}_results.jsonl"
    with open(output_file, "w") as f:
        for case in cases:
            result = {
                "id":     case.get("id"),
                "score":  case.get("score", 0),
                "passed": case.get("score", 0) >= 0.8
            }
            f.write(json.dumps(result) + "\n")

    print(f"Results saved to: {output_file}")

# call it 
run_eval_pipeline("eval_cases.jsonl", "results")

# ------ Sample i/p and o/p -------
# ------ Input — eval_cases.jsonl ---------
# {"id": "case_1", "score": 0.92}
# {"id": "case_2", "score": 0.55}
# {"id": "case_3", "score": 0.87}
# {"id": "case_4", "score": 0.40}
# {"id": "case_5", "score": 0.91}

# ------ What the code does to each case -------
# pythonresult = {
#     "id":     case.get("id"),
#     "score":  case.get("score", 0),
#     "passed": case.get("score", 0) >= 0.8
# }

# ----- Output file — eval_cases_results.jsonl -------
# jsonl{"id": "case_1", "score": 0.92, "passed": true}
# {"id": "case_2", "score": 0.55, "passed": false}
# {"id": "case_3", "score": 0.87, "passed": true}
# {"id": "case_4", "score": 0.4, "passed": false}
# {"id": "case_5", "score": 0.91, "passed": true}

# ----- Where the file is saved -----
# dataset.stem = "eval_cases"      ← filename without extension
# output file  = "eval_cases_results.jsonl"

# full path:
# results/
# └── eval_run/
#     └── 2024-11-15/              ← today's date
#         ├── inputs/
#         ├── outputs/
#         │   └── eval_cases_results.jsonl   ← file lands here
#         └── logs/

# ----- Terminal output when you run it 
# Eval run setup at: results/eval_run/2024-11-15
# Loaded 5 cases from eval_cases.jsonl
# Results saved to: results/eval_run/2024-11-15/outputs/eval_cases_results.jsonl
````
