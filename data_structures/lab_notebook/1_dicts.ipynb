```python
### What are Dicts? A dictionary is a python data structure that is a key-value store. Think of it like a JSON object where each data has a label.
    # Keys must be unique and hashable. 
        # Common key types are strings, numbers and tuples containing immutable(cannot be changed after creation)/hashable values.
        # Most AI code simply use String keys. 
        # If the same Key appears twice, the latter value wins. (See example 0. below)
    # Values can be anything : strings, numbers, boolean, lists or other dicts.

# Example 0 :
config = {
    "temperature": 0.7,
    "temperature": 0.2
}

print(config) # {"temperature": 0.2}

### Dicts Actions 
## 1. CREATING A DICT : Real AI Example - Model configuration(model_config) dictionary:
# Method A - Using curly braces {} : Standard and most used way
model_configA = {
    "model": "bert",
    "lr": 0.0003,
    "layers": [768, 768, 768],
    "tags": ["hallucination", "factual"]
    }
print("Creating dict - Method A : ", model_configA)

# Method B - Using dict() with named arguments
model_configB = dict(
    model="bert", 
    lr=0.0003, 
    layers=[768, 768, 768], 
    tags=["hallucination", "factual"]
)
print("Creating dict - Method B : ", model_configB)
# Python will interpret above as - model="bert" → "model": "bert" and lr=0.0003 → "lr": 0.0003

# Method C - Using dict() with key–value pairs (tuples)
model_configC = dict([
    ("model","bert"),
    ("lr",0.0003),
    ("layers",[768, 768, 768]),
    ("tags",["hallucination", "factual"])
    ]
)
print("Creating dict - Method B : ", model_configC)

## 2. READING VALUES FROM A DICT
# 2A - Direct access (KeyError if missing)
d = {"a": 1, "b": 2, "c": 3}
print(d["a"]) # o/p : 1
# print(d["e"]) # KeyError

# 2B - Safe access - .get()
print(d.get("a")) # o/p : 1
print(d.get("e")) # o/p : None
print(d.get("e", "Key is missing")) # o/p : Key is missing

# AI engineering habit: 
    # Use .get() for optional fields. 
    # Use direct access dict[key] when the field is required and its absence should throw an error.

## 3. ADD/UPDATE A DICT
score = {"red" : 20, "green" : 30, "orange" : 50} 
score["red"] = 100  # updating dict
score["yellow"] = 40 # adding to dict
print(score) # o/p : {'red': 100, 'green': 30, 'orange': 50, 'yellow': 40}

## 4. DELETE : pop() vs del
# pop(key): 
    # removes + returns value. Throws 'KeyError' if key is missing
# pop(key, default):
    # removes + returns value. Returns default if key is missing

val = score.pop("green")  #returns 30
print(val) 

print(score.pop("black", "None")) #safe delete. Returns None

# del : removes key, no returns and KeyError if missing
del score["red"]
# del score["green"] --> throws KeyError

## 5. ITERATING (this concept is used frequently)
model = {"model1" : "jupiter", "model2" : "mars", "model3" : "venues"}

for key in model:
    print(key)  # prints all the keys

for value in model.values():
    print(value) # prints all the values
       
for key, value in model.items():
    print(key, value) # prints each key and value in the dict

### 6. CHECKING MEMBERSHIP
    ## Membership means checking if something exists inside a collection. Python uses the operator 'in' and 'not in' to check for membership.
    ## By default Python checks for Keys, not Values in a dictionary.If we want to check values or full items then use .values() or .items()

student = {
    "name": "Sara",
    "age": 23,
    "city": "Boston"
}

print("name" in student) #True
print("DOB" in student) #False

print("Sara" in student.values()) #True
print("Rajvir" not in student.values()) #True

print(("age", 23) in student.items()) #True


### 7. Nested Dicts - Real world AI Resoponses look like this 

## In below example - 'api_response' is a dict and 'usage' and 'output' are nested dict inside it. 
api_response = {
    "id": "resp_001",
    "model": "claude-3",
    "usage": {
        "input_tokens": 120,
        "output_tokens": 85
    },
    "output": {
        "text": "The answer is 42",
        "finish_reason": "stop"
    }
}

# Accessing nested values
token = api_response["usage"]["input_tokens"] # 120
text = api_response["output"]["text"] # "The answer is 42"

# Safe nested access
token = api_response.get("usage",{}).get("input_tokens","unknown")

### 8. COMMON TRANSFORMATIONS: Merging Dicts and Update in-place
## Merging Dicts (creates a NEW dict, originals untouched)
defaults = {"temperature": 0.7, "max_tokens": 512}
overrides = {"temperature": 0.2, "model": "gpt-4"}

merged = defaults | overrides # Works for Python 3.9+ versions
print(merged) # o/p - {"temperature": 0.2, "max_tokens": 512, "model": "gpt-4"} 
# Basically Python reads and merges from left to right. So last one survives when there is a recurring key

merged = {**defaults, **overrides} #This merge option works for all python versions

## Update in-place (Does not create a new dict/object. Changes the existing one directly in memory.)
defaults.update(overrides)
print(defaults) # o/p: defaults = {"temperature": 0.2, "max_tokens": 512, "model": "gpt-4"} 

## Example of Real eval pipeline where update shines

# --- code in comment ------
# final_report = {} # Accumulating results across multiple eval runs
# for run in eval_runs:
#      final_report.update(run) # keeps adding/overwriting keys in one dict
# Creating a new dict each iteration would be wasteful in above example. Updating in-place is cleaner and more memory efficient here.
# --- code in comment ------

### 9. MOST USEFUL DICT METHODS
d = {"a": 1, "b": 2, "c": 3}

print(f"Returns all the keys in the dict: {d.keys()}") # dict_keys(['a', 'b', 'c'])

print(f"Returns all the values in the dict: {d.values()}") # dict_values([1, 2, 3])

print(f"Returns all (key + value) pairs in the dict: {d.items()}") # dict_items([('a', 1), ('b', 2), ('c', 3)])

print(f"Returns count of pairs in the dict : {len(d)}") # 3

### 10. Copying a Dict : Shallow copy and deep copy 
## In Python 'assignment' is not a 'Copy'. when you do b = a, you are not creating a new dict. 
## You are creating a second label pointing to the same dict in memory. See example below 

a1 = {"model": "gpt-4", "score": 0.9}
b1 = a1   # b1 is NOT a copy, both point to same dict

b1["score"] = 0.0             # changing b1...
print(a1["score"])            # 0.0 ← a1 got changed too!

## dict.copy() — Shallow Copy : Creates a new independent dict, but only at the top level.
## deepcopy(dict) — Copies everything — outer dict AND all nested dicts inside. Fully independent at every level.

a2 = {"model": "gpt-4", "usage": {"tokens": 100}}
b2 = a2.copy()

b2["model"] = "claude 3.5"
print(a2.get("model")) # gpt-4
print(b2.get("model")) # claude 3.5

b2["usage"]["tokens"] = 250
print(a2.get("usage").get("tokens")) # 250
print(b2.get("usage").get("tokens")) # 250

from copy import deepcopy  # deepcopy is not a built-in Python function. It lives inside Python's copy module and needs to be declared

a3 = {"model": "gpt-4", "usage": {"tokens": 100}}
b3 = deepcopy(a3)

b3["usage"]["tokens"] = 300
print(a3.get("usage").get("tokens")) # 100
print(b3.get("usage").get("tokens")) # 300


### 11. TWO PATTERNS - List of dicts (rows) and Dict of lists (columns) 

## List of dicts (like Rows): Why rows - Each dict represents one complete row — all fields for one record together.
results = [
    {"case_id": "case_1", "model": "gpt-4",   "score": 0.9},
    {"case_id": "case_2", "model": "claude",   "score": 0.7},
    {"case_id": "case_3", "model": "gpt-4",   "score": 0.8},
]
results[0]["model"]   # "gpt-4"

## Dict of lists (like Columns): Each key represents one complete column — one field across all records together.
results = {
    "case_id": ["case_1", "case_2", "case_3"],
    "model":   ["gpt-4",  "claude", "gpt-4"],
    "score":   [0.9,      0.7,      0.8],
}
results["model"][1]      # "claude"

### 12. REAL PATTERN EXAMPLE IN EVAL PIPELINES 

## Example 1 :
pred = {"id": 7, "label": "positive", "confidence": 0.91}

if pred.get("confidence", 0) >= 0.8:
    print("Keep:", pred["label"])

## Example 2 :
def evaluate_response(response: dict) -> dict:
    return {
        "passed": response.get("score", 0) >= 0.8,
        "model": response.get("model", "unknown"),
        "reason": response.get("reason", "no reason provided")
    }

### 13. UPDATE 
config = {"model": "gpt-4", "temperature": 0.7}
result = config.update({"temperature": 0.2, "max_tokens": 512})
print(result) # O/P - None
print(config) # {"model": "gpt-4", "temperature": 0.2, "max_tokens": 512}

## Why? - config.update() does two things:
        ### Modifies config directly in-place ✅
        ### Returns nothing — which in Python means it returns None

### 14. OTHER IMP DICT USAGE IN THE AI WORLD

    ## 14.1 Dictionaries as fast lookup tables
        # Think of dicts not only as "JSON-like Objects", but also as FAST MAPPING/LOOK-UP TABLES.
        # That's very important for AI evaluation systems.

    ## 14.2 Counting / aggregation pattern
        # Pattern worth undersanding : counts[key] = counts.get(key, 0) + 1   
    
labels = ["pass", "fail", "pass", "unsafe", "fail"]

counts = {}
for label in labels:
    counts[label] = counts.get(label,0) + 1

print(counts) 

## Result :
# {
#     "pass": 2,
#     "fail": 2,
#     "unsafe": 1
# }

    ## 14.3 Grouping results using dictionaries
        # This is a very real evaluation-engineering pattern: group results by model, dataset, grader, category, failure type, etc.

results = [
    {"model": "gpt-4", "score": 0.9},
    {"model": "claude", "score": 0.8},
    {"model": "gpt-4", "score": 0.85}
] 
# Here data is organized as a list of dictionaries.
# Now if we want the answer : What are all the scores for each model? i.e I want below : 
#      scores_by_model = {
#     "gpt-4": [0.9, 0.85],
#     "claude": [0.8]
# }

# Here's how to get it :
scores_by_model = {} 

for result in results: # result is a dict in each iteration like result = {"model": "gpt-4", "score": 0.9}
    model = result["model"]
    score = result["score"]

if model not in scores_by_model:
    scores_by_model[model]= []

scores_by_model[model].append(score)


    ## 14.4 Safe nested access has one subtle trap
# Case 1 : 
api_response = {
    "usage": {
        "input_tokens": 120
    }
}

# Case 2 :
api_response = {
    "model": "gpt-4"
}

# Case 3 : 
api_response = {
    "usage": None
}

# Solution 1 : This works for Case 1 and 2 but for Case 3 throws error - AttributeError: 'NoneType' object has no attribute 'get'
token = api_response.get("usage", {}).get("input_tokens", "unknown")

# Solution 2 : Works for all 3 cases and does error handeling
usage = api_response.get("usage") or {}
token = usage.get("input_tokens", "unknown")



```

```python
# AI Eval practice set for Dict

## Q1. Parse an LLM Evaluation Response

# Write code that:
# - Prints the model name
# - Safely reads "reason" with default value "not provided"
# - Reads 'latency_ms'
# - Checks whether "score" exists in the dictionary
# - Prints "High confidence result" if score is >= 0.9

response = {
    "model": "gpt-4",
    "score": 0.91,
    "passed": True,
    "metadata": {
        "latency_ms": 820,
        "tokens": 145
    }
}

print(f"The model name is: {response.get("model","No name found")}")
model_reason = response.get("reason", "not provided")
model_latency = response.get("metadata",{}).get("latency_ms", None)

if("score" in response):
    if(response["score"] >= 0.9):
        print("High confidence result")

# ----------------------------------------------------------------------
## Q2. Aggregate Failure Counts

# An AI safety evaluation produced:
failures = [
    "hallucination",
    "timeout",
    "hallucination",
    "unsafe",
    "timeout",
    "hallucination"
]

# Create 'failure_counts' so that it becomes - 
# {
#     "hallucination": 3,
#     "timeout": 2,
#     "unsafe": 1
# }

failure_counts = {}

for failure in failures:
        failure_counts[failure] = failure_counts.get(failure,0) + 1
    
print(failure_counts)

# ----------------------------------------------------------------------
## Q3. Group Scores by Model

results = [
    {"model": "gpt-4", "score": 0.91},
    {"model": "claude", "score": 0.88},
    {"model": "gpt-4", "score": 0.95},
    {"model": "gemini", "score": 0.82},
    {"model": "claude", "score": 0.90}
]

# Create 'scores_by_model' so the final result becomes :
# {
#     "gpt-4": [0.91, 0.95],
#     "claude": [0.88, 0.90],
#     "gemini": [0.82]
# }

# Requirements:
# - Loop through results
# - Do not hardcode model names
# - Create a list when a model appears for the first time
# - Append later scores to the correct model

scores_by_model = {}

for result in results:
    model = result["model"]
    score = result["score"]

    if model not in scores_by_model:
        scores_by_model[model] = []

    scores_by_model[model].append(score)

print(f"Scores by Model: {scores_by_model}")
# Scores by Model: {'gpt-4': [0.91, 0.95], 'claude': [0.88, 0.9], 'gemini': [0.82]}

# ----------------------------------------------------------------------
## Q4. Safely Process Inconsistent API Responses

responses = [
    {
        "id": "case_1",
        "usage": {
            "input_tokens": 120
        }
    },
    {
        "id": "case_2"
    },
    {
        "id": "case_3",
        "usage": None
    }
]

# Write code that prints:
# case_1 -> 120
# case_2 -> unknown
# case_3 -> unknown

tokens_usage_by_id = {}

for response in responses:
     case_id = response["id"]

     tokens_usage_by_id[case_id] = (response.get("usage") or {}).get("input_tokens", "unkown")

print("********************************************")
print(f"tokens_usage_by_id  is: {tokens_usage_by_id}")     

# ----------------------------------------------------------------------
# Q5. Build an Eval Summary Report

eval_results = [
    {
        "case_id": "T001",
        "model": "gpt-4",
        "category": "hallucination",
        "score": 0.92,
        "passed": True,
        "usage": {"tokens": 140}
    },
    {
        "case_id": "T002",
        "model": "claude",
        "category": "safety",
        "score": 0.76,
        "passed": False,
        "usage": {"tokens": 110}
    },
    {
        "case_id": "T003",
        "model": "gpt-4",
        "category": "hallucination",
        "score": 0.81,
        "passed": True,
        "usage": None
    },
    {
        "case_id": "T004",
        "model": "claude",
        "category": "safety",
        "score": 0.69,
        "passed": False
    },
    {
        "case_id": "T005",
        "model": "gemini",
        "category": "reasoning",
        "score": 0.88,
        "passed": True,
        "usage": {"tokens": 160}
    }
]

# Build this dynamically using Example below:

# summary = {
#     "total_cases": 5,
#     "passed_cases": 3,
#     "failed_cases": 2,

#     "failures_by_category": {
#         "safety": 2
#     },

#     "scores_by_model": {
#         "gpt-4": [0.92, 0.81],
#         "claude": [0.76, 0.69],
#         "gemini": [0.88]
#     },

#     "tokens_by_model": {
#         "gpt-4": 140,
#         "claude": 110,
#         "gemini": 160
#     }
# }

# Rules:
# - total_cases = total records
# - Increment passed/failed counts
# - Only failed records contribute to failures_by_category
# - Group scores by model
# - Sum tokens by model
# - Missing or None usage means 0 tokens
# - Do not hardcode model/category names
# - No comprehensions yet

summary = {}
summary["total_cases"] = len(eval_results)
summary["passed_cases"] = 0
summary["failed_cases"] = 0
failures_by_category = {}
scores_by_model = {}
tokens_by_model = {}

for result in eval_results:

     if(result.get("passed") == True):
        summary["passed_cases"] +=1

     else:
        summary["failed_cases"] +=1

        category = result["category"]
        if(category not in failures_by_category):
            failures_by_category[category] = 1
        else:
          failures_by_category[category] += 1
            
     model = result.get("model", "unknown")
     if(model not in scores_by_model):
         scores_by_model[model] = []
         
     scores_by_model[model].append(result.get("score",0))    

     if(model not in tokens_by_model):
         tokens_by_model[model] = (result.get("usage") or {}).get("tokens",0)
     else:
        tokens_by_model[model] += (result.get("usage") or {}).get("tokens",0)   

summary["failures_by_category"] = failures_by_category
summary["scores_by_model"] = scores_by_model
summary["tokens_by_model"] = tokens_by_model

print(f"The Eval Summary Report: {summary}")

# ----------------------------------------------------------------------
# Q6. Find and Store Failed Eval Cases - Combines Lists and Dicts

eval_results = [
    {
        "case_id": "T001",
        "category": "reasoning",
        "score": 0.91,
        "passed": True
    },
    {
        "case_id": "T002",
        "category": "hallucination",
        "score": 0.55,
        "passed": False
    },
    {
        "case_id": "T003",
        "category": "safety",
        "score": 0.48,
        "passed": False
    },
    {
        "case_id": "T004",
        "category": "reasoning",
        "score": 0.87,
        "passed": True
    },
    {
        "case_id": "T005",
        "category": "hallucination",
        "score": 0.62,
        "passed": False
    }
]

# Create a new list - ;failed_cases' containing only failed records.
# Then also : 
    # Print total failed cases
    # Print the first failed case_id
    # Print the most recent failed case_id using negative indexing
    # Create a dictionary called failed_case_ids_by_category

failed_cases = []  # contains all failed records

for result in eval_results:
    if(result.get("passed", "unknown") == False):
        failed_cases.append(result)

print(f"Total failed cases is: {len(failed_cases)}") # Prints total failed cases

first_failed_caseid = failed_cases[0].get("case_id","Case id doesnot exit")
print(f"The first failed case_id is: {first_failed_caseid}") # Prints the first failed case_id

most_recent_failed_caseid = failed_cases[-1].get("case_id","Case id doesnot exit")
print(f"The most recent failed case_id is: {most_recent_failed_caseid}") # Prints the most recent failed case_id

failed_case_ids_by_category = {} 

for case in failed_cases:
    category = case.get("category", "unknown")
    case_id = case.get("case_id", "unknown" )

    if(category not in failed_case_ids_by_category):
        failed_case_ids_by_category[category] = []

    failed_case_ids_by_category[category].append(case_id)    

print(f"Details of 'failed_case_ids_by_category' is: {failed_case_ids_by_category}")
print("********************************************")

# ----------------------------------------------------------------------
# Q7. Build a Model Comparison Report - This problem combines lists, dicts, nested structures, grouping, aggregation, missing values, membership, and filtering.

results = [
    {
        "case_id": "T001",
        "model": "gpt-4",
        "category": "reasoning",
        "score": 0.91,
        "passed": True,
        "usage": {"tokens": 120}
    },
    {
        "case_id": "T002",
        "model": "claude",
        "category": "reasoning",
        "score": 0.89,
        "passed": True,
        "usage": {"tokens": 105}
    },
    {
        "case_id": "T003",
        "model": "gpt-4",
        "category": "safety",
        "score": 0.62,
        "passed": False,
        "usage": {"tokens": 150}
    },
    {
        "case_id": "T004",
        "model": "claude",
        "category": "safety",
        "score": 0.78,
        "passed": False,
        "usage": None
    },
    {
        "case_id": "T005",
        "model": "gemini",
        "category": "reasoning",
        "score": 0.84,
        "passed": True
    },
    {
        "case_id": "T006",
        "model": "gpt-4",
        "category": "hallucination",
        "score": 0.71,
        "passed": False,
        "usage": {"tokens": 130}
    }
]

# Build 'model_report' strtucture dynamically.

# Rules:
# - Do not hardcode model names
# - When you encounter a model for the first time, initialize its nested dictionary
# - Add every score to that model's scores
# - Increment either passed or failed
# - If failed, save that test's case_id
# - Sum token usage
# - Missing "usage" or "usage": None means 0 tokens
# - Do not use comprehensions
# - Do not modify the original results


model_report = {}

for result in results:
    model = result.get("model","unknown")

    if(model not in model_report):
        model_report[model] = {
            "scores":[],
            "passed":0,
            "failed":0,
            "failed_case_ids":[],
            "total_tokens":0
        }

    model_report[model]["scores"].append(result.get("score"))   

    if(result.get("passed") == True):
        model_report[model]["passed"] += 1
    else:    
        model_report[model]["failed"] += 1
        model_report[model]["failed_case_ids"].append(result.get("case_id"))      
    
    model_report[model]["total_tokens"] += (result.get("usage") or {}).get("tokens",0)
   
print(f"The original 'results is: {results}")
print(f"The model comparison report is: {model_report}")

# The model comparison report is: {'gpt-4': {'scores': [0.91, 0.62, 0.71], 'passed': 1, 'failed': 2, 'failed_case_ids': ['T001', 'T003', 'T006'], 'total_tokens': 400}, 
#                                  'claude': {'scores': [0.89, 0.78], 'passed': 1, 'failed': 1, 'failed_case_ids': ['T002', 'T004'], 'total_tokens': 105}, 
#                                  'gemini': {'scores': [0.84], 'passed': 1, 'failed': 0, 'failed_case_ids': ['T005'], 'total_tokens': 0}}
```
