```python
### Sets : 
# - It is an unordered collection of unique items. So no duplicates allowed . 
# - Sets are not indexable or sliceable.
# - Plus fast membership check (using 'in' and 'not in' due to using hashing and not linear search like Lists does)
    # - list membership → typically O(n)
    # - set membership  → typically O(1) average

# Difference b/w Lists and Sets :
    # 1. List → order matters → indexing/slicing supported
    # 2. Set  → membership/uniqueness matters → no indexing/slicing

### Part 1 : Creating Sets
## Default Syntax : 
tags = {"hallucination", "factual", "refused"}

## From a list — great for deduplication
scores   = [0.9, 0.4, 0.9, 0.85, 0.4]
unique   = set(scores)
print(unique)    # {0.9, 0.4, 0.85}  ← duplicates removed

# Empty set
ids = set()  # Donot do 'ids = {}' --> This will create a dict

### Part 2 : Adding items to Sets
s = set()
s.add(4)                # add single element only
s.update({5, 6, 7})   # or update([5, 6, 7]) works too . update is used to add multiple elements
print(s)  # {4, 5, 6, 7}

### Part 3 : Removing items from Sets
d = {10,20,30,40,50,60}
d.remove(20)            # Not safe remove as gives error if element not present
d.discard(100)           # safe (no error)
print(d.pop())          # removes and returns random element
d.clear()               # removes all elements

### Part 4 : Checking Membership
fruit = {"apple", "mango", "grapes", "orange"}
"apple" in fruit    # True
"banana" not in fruit # True

### Part 5 : Set Operations — The Real Power
a = {1, 2, 3}
b = {3, 4, 5}

## Union (combine all unique elements in both sets )
a | b  # {1, 2, 3, 4, 5}

## Intersection (combine only items in BOTH sets)
a & b  # {3}

## Difference (items in first but NOT in second )
a - b  # {1, 2}

## Symmetric Difference (Items that exist in either set, but NOT in both.)
a ^ b  # {1, 2, 4, 5}

## Subset and Superset
required_tags = {"hallucination", "factual"}
response_tags = {"hallucination", "factual", "refused", "grounded"}

# Is required_tags fully contained inside response_tags?
required_tags.issubset(response_tags)      # True
required_tags <= response_tags             # same thing, cleaner

# Does response_tags contain everything in required_tags?
response_tags.issuperset(required_tags)    # True
response_tags >= required_tags             # same thing

### Part 6 : Iterating through Sets
for x in s:
    print(x)

### Part 7 : Common Patterns (AI Use Cases)

## Deduplication Pattern — Used Constantly
raw_ids = ["case_1", "case_2", "case_1", "case_3", "case_2"]
unique_ids = list(set(raw_ids)) # Instant deduplication
print(unique_ids)    # ["case_1", "case_2", "case_3"]

    # Note : 
    # Deduplication removes duplicates correctly, but does not guarantee that the original order is preserved. 
    # So use when order doesn't matter.

## Fast filtering
tokens = ["a", "b", "a", "on"]
stopwords = {"the", "is", "on"}

filtered = [w for w in tokens if w not in stopwords] # ** See details of this below
print(filtered) # ['a', 'b', 'a']

## [w for w in tokens if w not in stopwords] - [expression for item in collection if condition]
#  w : what to put in the list (expression)
#  for w in tokens : loop
#  if w not in stopwords : condition

### Part 8 : Important Notes
## Set elements must be hashable
    # - A set itself is mutable (can be changed after creation), but the items inside it must be hashable. 

# Hashable : Python can give the object a stable identity-like hash value. Sets use a hash table internally to find items quickly.
    # For an object to be safely used inside a set, its hash must not change while it's inside the set.
    # Hence Strings, numbers, booleans, tuples usually work as set elements. Lists and dicts do not.      

##  .isdisjoint() — very useful for validations
    # This tells us whether two sets have zero overlap.

unsafe_tags = {"violence", "self_harm", "hate"}
response_tags1 = {"factual", "helpful"}

unsafe_tags.isdisjoint(response_tags1) # True

response_tags2 = {"helpful", "violence"}
unsafe_tags.isdisjoint(response_tags2) # False

## Very important AI-eval pattern: expected vs actual
    # This is probably the most useful practical Set pattern after deduplication.

expected = {"grounded", "factual", "safe"}
actual = {"grounded", "safe", "verbose"}

missing = expected - actual # What's missing?
print(missing) # {"factual"}

unexpected = actual - expected # What unexpected labels appeared?
print(unexpected) # {"verbose"}

matched = expected & actual # What matched?
print(matched) # {"grounded", "safe"}

# Find total items in a set : len(set_name)
```

```python
# Set Data Structure Practice — 5 AI Eval Problems

## Problem 1 — Deduplicate failure categories

# An eval run produced duplicate failure labels:

failure_labels = [
    "hallucination",
    "timeout",
    "hallucination",
    "unsafe",
    "timeout",
    "tool_error",
    "hallucination"
]

# Create a set called 'unique_failures' that contains only the unique failure types.
# Then:
# - print unique_failures
# - print the number of unique failure categories
# - check whether "unsafe" occurred
# - check whether "bias" did not occur
# Do not manually hardcode the final set.

unique_failures = set(failure_labels)
print(f"The unique failures are: {unique_failures}")
print(f"The number of unique failure categories are: {len(unique_failures)}")

if("unsafe" in unique_failures):
    print("unsafe occured")
else:
    print("unsafe didnot occur")  

if("bias" in unique_failures):
    print("bias occured")
else:
    print("bias didnot occur")  

print("********************************")

#-------------------------------------------------------------
## Problem 2 — Expected vs actual safety labels

# An evaluator expects these labels:
expected_labels = {
    "safe",
    "grounded",
    "factual",
    "helpful"
}

# The model actually received:
actual_labels = {
    "safe",
    "grounded",
    "verbose",
    "helpful"
}

# Using set operations, create:
    # missing_labels      → expected but not actually present
    # unexpected_labels   → actual but not expected
    # matched_labels      → present in both

missing_labels = expected_labels - actual_labels
unexpected_labels = actual_labels - expected_labels
matched_labels = expected_labels & actual_labels

print(f"Missing labels: {missing_labels}")
print(f"Unexpected labels: {unexpected_labels}")
print(f"Matched labels: {matched_labels}")

print("********************************")

#-------------------------------------------------------------
## Problem 3 — Validate required eval coverage

# Your benchmark requires these evaluation categories:
required_categories = {
    "reasoning",
    "safety",
    "hallucination",
    "tool_use"
}

# A particular benchmark run covered:
tested_categories = {
    "reasoning",
    "safety",
    "hallucination",
    "tool_use",
    "coding",
    "retrieval"
}

# Check whether all required categories were covered
if(required_categories.issubset(tested_categories)):
    print("All required categories were covered")
else:
    print("All required categories were not covered")    

# Find any extra categories that were tested
extra_categories = tested_categories - required_categories
print(f"Extra categories tested are: {extra_categories}")
print("********************************")

#-------------------------------------------------------------
## Problem 4 - Detect prohibited labels

# A safety evaluation defines:
prohibited_labels = {
    "violence",
    "self_harm",
    "hate",
    "illegal_activity"
}

# The model response was tagged with:
response_labels = {
    "helpful",
    "factual",
    "violence",
    "verbose"
}

# 1. Determine whether response_labels and prohibited_labels are completely disjoint
if(response_labels.isdisjoint(prohibited_labels)):
    print("'response_labels' and 'prohibited_labels' are completely disjoint")
else:
    print("'response_labels' and 'prohibited_labels' are NOT completely disjoint")    

# 2. Create a set containing the prohibited labels actually detected. 
# Print - "Safety violation detected" if there is any overlap. Prints the violating labels

detected_prohibited_labels = response_labels & prohibited_labels
if(detected_prohibited_labels): # Python treats all empty containers as False and non-empty ones as True
    print(f"Safety violation detected. Details: {detected_prohibited_labels}")
else:
    print("No safety violation detected")    

print("********************************")

#-------------------------------------------------------------
## Problem 5 - Compare two model eval runs

# Two models were evaluated on the same benchmark.
model_a_failures = {
    "T002",
    "T005",
    "T008",
    "T011"
}

model_b_failures = {
    "T005",
    "T008",
    "T009",
    "T014"
}

# Create sets for:
    # failed_by_both
    # only_model_a_failed
    # only_model_b_failed
    # failed_by_exactly_one_model

failed_by_both = model_a_failures & model_b_failures
only_model_a_failed = model_a_failures - model_b_failures
only_model_b_failed = model_b_failures - model_a_failures
failed_by_exactly_one_model = model_a_failures ^ model_b_failures

# Then answer programmatically:
# - How many cases did both models fail?
total_cases = len(failed_by_both)
print(f"Cases both models failed: {total_cases} ")

# - How many total unique test cases failed across either model?
unique_cases = model_a_failures | model_b_failures
print(f"Unique cases both models failed: {len(unique_cases)} ")

# Or another way : 
    # unique_cases = set()
    # unique_cases.update(failed_by_both,only_model_a_failed,only_model_b_failed)

print("********************************")

#--------------------------------------------------------------------------------------
## Problem 6 - Build a failure-analysis report  (Combined List + Dict + Set Problems)
eval_results = [
    {
        "case_id": "T001",
        "model": "gpt-4",
        "category": "reasoning",
        "passed": True
    },
    {
        "case_id": "T002",
        "model": "gpt-4",
        "category": "hallucination",
        "passed": False
    },
    {
        "case_id": "T003",
        "model": "claude",
        "category": "safety",
        "passed": False
    },
    {
        "case_id": "T004",
        "model": "claude",
        "category": "reasoning",
        "passed": True
    },
    {
        "case_id": "T005",
        "model": "gpt-4",
        "category": "safety",
        "passed": False
    },
    {
        "case_id": "T006",
        "model": "claude",
        "category": "hallucination",
        "passed": False
    }
]

# [1] Build 'failed_cases = []' containing the full dictionaries for failed cases.
# [2] Then create 'unique_failed_categories = set()' containing the unique categories where failures occurred.

failed_cases = []
unique_failed_categories = set()

for result in eval_results:
    if(result.get("passed") is False):
        failed_cases.append(result)
        unique_failed_categories.add(result.get("category", "unknown"))

print(f"'failed_cases' containing full dict for failed cases: {failed_cases}'")

# [3] Now create 'failures_by_model = {}' so that the final structure becomes:
    # {
    #     "gpt-4": {"T002", "T005"},
    #     "claude": {"T003", "T006"}
    # }

failures_by_model = {}
for case in failed_cases:
    model = case.get("model", "unknown")
    case_id = case.get("case_id", "unknown")

    if(model not in failures_by_model):
        failures_by_model[model] = set()

    failures_by_model[model].add(case_id)

print(f"Failures by model is: {failures_by_model}")

# Finally print:
    # total number of failed records
print(f"Total number of failed records: {len(failed_cases)}")

    # unique failure categories
print(f"Total number of unique failure categories: {len(unique_failed_categories)}")
print(f"Unique categories where failures occurred: {unique_failed_categories}'")

print("********************************")

#--------------------------------------------------------------------------------------
## Problem 7 — Harder: Cross-model benchmark comparison (Combined List + Dict + Set Problems)

# You receive :
results = [
    {"case_id": "T001", "model": "gpt-4", "passed": True},
    {"case_id": "T002", "model": "gpt-4", "passed": False},
    {"case_id": "T003", "model": "gpt-4", "passed": False},
    {"case_id": "T004", "model": "gpt-4", "passed": True},

    {"case_id": "T001", "model": "claude", "passed": True},
    {"case_id": "T002", "model": "claude", "passed": False},
    {"case_id": "T003", "model": "claude", "passed": True},
    {"case_id": "T004", "model": "claude", "passed": False}
]

# Build this dynamically:
#   failures_by_model = {
#       "gpt-4": {"T002", "T003"},
#       "claude": {"T002", "T004"}
#   }

failures_by_model = {}
for result in results:
    model = result.get("model", "unknown")
    case_id = result.get("case_id", "unknown")

    if(model not in failures_by_model):
        failures_by_model[model] = set()

    if(result.get("passed") is False):
        failures_by_model[model].add(case_id)

# Now use the sets inside the dictionary to determine:
gpt4_failures = failures_by_model.get("gpt-4", set())
claude_failures = failures_by_model.get("claude", set())

    # failed_by_both
failed_by_both = (gpt4_failures) & (claude_failures)
print(f"Failed by both is: {failed_by_both}")

    # only_gpt4_failed
only_gpt4_failed = (gpt4_failures) - (claude_failures)
print(f"Only GPT4 failed: {only_gpt4_failed}")

    # only_claude_failed
only_claude_failed = (claude_failures) - (gpt4_failures)
print(f"Only claude failed: {only_claude_failed}")

    # all_failed_cases    
all_failed_cases = (gpt4_failures)| (claude_failures)
print(f"All failed cases: {all_failed_cases}")















```
