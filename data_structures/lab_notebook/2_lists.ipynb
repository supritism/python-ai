### Introduction - Lists : 
- Lists are an ordered, mutable collection of items. Three key points : 
  - Each item in the list has an index.
  - Lists allow duplicate items. 
  - Lists are mutable means we can change/update their contents after creating them. Syntax : list_name[index] = new_value
  Example : 
    scores = [1,2,3] 
    scores[1] = 4   

Example :
models = ["gpt-4", "claude-3", "gemini"]

0 -> gpt-4
1 -> claude-3
2 -> gemini

In Python we can create a list with mixed data types but Production AI code avoids them. Keep lists uniform so you can safely loop, sort, and compute over them.

## Part 10: Slicing + Negative Index Slicing + Step Slicing - V.V Imp
- Slicing means extracting portions of a list.
- Slicing returns a brand-new list — it doesn't touch the original list.
- Rule of slicing : list[start : stop] — start is index to begin from (inclusive), stop is index to end at (exclusive — this item is NOT included)

- Step slicing : Slicing a list with a third parameter 'step' that controls how many items to jump each time.
- list[start:stop:step]
- For step remember default is +1 which is left to right moving forward. -1 is from right to left moving backwards

## Part 13: Common Beginner Mistakes
- Mistake 1 — Using .sort() and expecting a return value. sort() returns None, not the list.
- Mistake 2 — Modifying a list while looping over it. Never modify list you're iterating.
- Mistake 3 - Mistake 3 — append vs extend confusion. 
  - scores = [0.9, 0.4]
  - scores.append([0.85, 0.6])    # adds list AS ONE ITEM
  - print(scores)    # [0.9, 0.4, [0.85, 0.6]]  ← nested list!
  
  - scores.extend([0.85, 0.6])    # adds items individually
  - print(scores)    # [0.9, 0.4, 0.85, 0.6]  ← flat, what you wanted

```python
## Part 2: Accessing items in a List - Indexing
# List are 0-indexed
fruits = ["apple", "mango" , "banana"]
print(fruits[0]) # apple

# Negative Indexing : Counts from the end starting with -1 for the last element in list
print(fruits[-1]) #banana

 #           0        1         2
fruits = ["apple", "mango" , "banana"]
#           -3        -2        -1

#--------------------------------------------#
## Part 3: Adding items to a List - append, extend and insert
model_rating = ["good", "average", "bad"]

model_rating.append("unexpectable")
print(model_rating) # ['good', 'average', 'bad', 'unexpectable']

model_rating.extend(["bar-raising", "broken"])
print(model_rating) # ['good', 'average', 'bad', 'unexpectable', 'bar-raising', 'broken']

model_rating.insert(3,"poor")
print(model_rating) # ['good', 'average', 'bad', 'poor', 'unexpectable', 'bar-raising', 'broken']

#--------------------------------------------#
## Part 4: Updating a List
scores = [10,20,30]
scores[1] = 25
print(scores) # [10, 25, 30]

#--------------------------------------------#
## Part 5: Removing items from a List
models = ["gpt-4", "mistral", "claude-3", "gemini", "mistral"]

models.remove("mistral") #remove by value. 1st occurence
print(models) #['gpt-4', 'claude-3', 'gemini', 'mistral']

model_removed = models.pop(1) #remove by index
print(model_removed) #claude-3
print(models) #['gpt-4', 'gemini', 'mistral']
models.pop()
print(models) #['gpt-4', 'gemini']

models_removed = models.pop() #removed the last item in the list
print(model_removed) #gemini
print(models) #['gpt-4']

failures = ["timeout", "hallucination", "unsafe", "tool_error","compliance"]
del failures[1]
print(failures) # ["timeout", "unsafe", "tool_error","compliance"]

del failures[1:3] # slicing syntax [start:stop]
print(failures) # ["timeout", "compliance"]]

# del failures[:2]    # delete first 2 items
# del failures[2:]    # delete index 2 through the end
# del failures[-2:]   # delete last 2 items

models.clear()
print(models) #[]

#--------------------------------------------#
## Part 6: Searching and Checking Membership 
scores = [0.9, 0.4, 0.85, 0.6, 0.4, 0.6]

# Check membership
print(0.9 in scores) # True
print(1.0 not in scores) # True

# Get index of first occurence of an item
print(scores.index(0.6)) # 3

# Get total occurence/s of an item in list
print(scores.count(0.6)) # 2

# Get length of list
print(len(scores)) # 6

#--------------------------------------------#
## Part 7: Iterating through Lists/ Looping
scores = [0.9, 0.4, 0.85]

# Simple loop
for score in scores:
    print(score)

# Use enumerate() to loop with index — when you need position too
for index,score in enumerate(scores):
    print(f"Counter: {index} - {score}")

# Loop with counter starting from 1
for index,score in enumerate(scores, start=1):
    print(f"Counter: {index} - {score}")

#--------------------------------------------#
## Part 8: Sorting
scores = [0.9, 0.4, 0.85, 0.6, 1.0]

# .sort()
scores.sort(reverse=True) # Sorts in descending order
print(scores) # [1.0, 0.9, 0.85, 0.6, 0.4]
print(scores.sort()) # None as .sort returns nothing
print(scores) # [0.4, 0.6, 0.85, 0.9, 1.0]

# sorted()
sorted_scores = sorted(scores)
print(sorted(sorted_scores, reverse=True)) # [1.0, 0.9, 0.85, 0.6, 0.4]
#--------------------------------------------#
## Part 9: Copy (Same trap as Dict)
a = [0.9, 0.4, 0.85]
# Assignment — NOT a copy
b = a
b.append(0.6)
print(a)    # [0.9, 0.4, 0.85, 0.6]  ← 'a' got modified too!
# Why? - Because 'a' and 'b' are name tags (variables) bound to the same object. b = a doesnot produce a new object. Check below :
    # id(<variable>) gives us the unique ID of the object a variable is pointing to.
print(f"ID of object 'a' is pointint to : {id(a)}") # 4510854080
print(f"ID of object 'b' is pointint to : {id(b)}") # 4510854080

# Shallow copy — safe for flat lists
b = a.copy()   # update to 'b' will not affect 'a'

# Deep copy — for nested lists
from copy import deepcopy
b = deepcopy(a)

#--------------------------------------------#
## Part 10: Slicing + Negative Index Slicing + Step Slicing - V.V Imp
scores =    [0.9, 0.4, 0.85, 0.6, 1.0]
#index :      0    1    2     3    4
#-ve index :  -5   -4   -3   -2   -1

# Slicing
print(scores[1:3])   # From index 1 up to index 3 - [0.4, 0.85]

print(scores[:3])   # From start up to index 3 - [0.9, 0.4, 0.85]

print(scores[2:])   # From index 2 to the end - [0.85, 0.6, 1.0]

print(scores[:])  # Entire list : Similar to Shallow copy -  [0.9, 0.4, 0.85, 0.6, 1.0]

# Negative Index slicing
print(scores[-2:])   # Last 2 items - [0.6, 1.0]

print(scores[-3:])   # Last 3 items - [0.85, 0.6, 1.0]

print(scores[:-1])   # Everything except last item - [0.9, 0.4, 0.85, 0.6]

print(scores[:-2])   # Everything except last 2 - [0.9, 0.4, 0.85]

# Step Slicing
print(f"Step slicing : {scores[::2]}")  # Every other item (step=2) - [0.9, 0.85, 1.0]

print(f"Step slicing : {scores[::3]}")   # Every 3rd item - [0.9, 0.6]

# 'Step = 1' means jump by 1 — land on every position, skip nothing.
# 'Step = 2' means jump by start/previous index + 2 for next item. [0 index value, 2 IV, 4 IV, 6 IV ....]
# 'Step = 3' means jump by start/previous index + 3 for next item. [0 index value, 3 IV, 6 IV, 9 IV ....]

#--------------------------------------------#
## Part 11: Reverse a List
# Way 1 — .reverse() in-place
scores = [0.9, 0.4, 0.85, 0.6, 1.0]
scores.reverse()
print(scores) # [1.0, 0.6, 0.85, 0.4, 0.9]  ← original changed

# Way 2 - [::-1] slice
scores = [0.9, 0.4, 0.85, 0.6, 1.0]
reversed_scores = scores[::-1]
print(reversed_scores)  # [1.0, 0.6, 0.85, 0.4, 0.9]  ← new list
print(scores) # [0.9, 0.4, 0.85, 0.6, 1.0]
# For step remember -1 always starts backward moving right to left.

#--------------------------------------------#
## Part 12: Summarizing commonly used List Functions (V.V Imp)
scores = [0.9, 0.4, 0.85]

scores.copy()             # shallow copy
len(scores)               # number of items
sorted(scores)          # new sorted list, original safe
min(scores)               # smallest value in list
max(scores)               # largest value in list
sum(scores)               # sum of all values in list
scores.append(0.7)        # add one item to end
scores.extend([0.6, 1.0]) # add multiple items to end
scores.insert(0, 0.5)     # insert at position
scores.remove(0.4)        # remove by value
scores.pop()              # remove and return last item
scores.pop(0)             # remove and return item at index
scores.sort()             # sort in-place, returns None
scores.reverse()          # reverse in-place, returns None
scores.index(0.9)         # find index of first occurence of value
scores.count(0.9)         # count occurrences
scores.clear()            # empty the list

#--------------------------------------------#








```
