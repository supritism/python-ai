```python
### What is a Tuple? 
# - Ordered, immutable collection of elements
    # - Tuple immutability means tuple elements cannot be added, removed, or reassigned. 
    # However, a tuple may contain mutable objects, and those inner objects can still change. (See Example 0 below)
# - Allows duplicates + Indexed (like lists)

# Example 0
    # t = ("gpt-4", 0.9)
    # t[0] = "claude"   # ❌ TypeError

t = ("gpt-4", [0.9, 0.8])
t[1].append(0.7)
print(t) # ("gpt-4", [0.9, 0.8, 0.7])


### 1. CREATING A TUPLE
# default
t = ("gpt-4", 0.9, True)

# without brackets — still a tuple (Packing)
t = "gpt-4", 0.9, True

# single item — trailing comma is REQUIRED
t = ("gpt-4",)    # ✅ tuple
t = ("gpt-4")     # ❌ just a string

# empty tuple
t = ()

# from a list
t = tuple([0.9, 0.4, 0.85]) # tuple() function takes any iterable (like list, string, set) and converts it into a tuple.

### 2. ACCESSING A TUPLE + IMMUTABILITY + ADDING ITEMS TO TUPLE
t = ("gpt-4", 0.9, True, "stop")

t[0]       # "gpt-4"   — first item
t[-1]      # "stop"    — last item
t[1:3]     # (0.9, True) — slice rules from Lists apply here

# membership check
0.9 in t   # True
len(t)     # 4

# ❌ cannot modify - Immutability
# t[0] = "claude"   # TypeError
# t.append("x")     # AttributeError

# ADDING Items to Tuples
    # Tuples are immutable, so we cannot use -  t.append(...) or directly modify them.
    # Instead, we create a new tuple by concatenating tuples using + or +=.

t = (1, 2)
t = t + (3,)

print(t)
# (1, 2, 3)

# Adding a Tuple Inside Another Tuple
compressed = ()

compressed += (("pass", 2),)
compressed += (("fail", 3),)

# O/P - 
# (
#     ("pass", 2),
#     ("fail", 3)
# )

### 3. TUPLE PACKING 
t = (1, 2, 3)         # packing
v = "gpt-4", 0.9, True # packing

### 4. TUPLE UNPACKING
t = ("gpt-4", 0.9, True)

# basic — must match exact count
model, score, passed = t

# * captures the rest
scores = (0.9, 0.4, 0.85, 0.6, 1.0)
first, *middle, last = scores
# first  → 0.9
# middle → [0.4, 0.85, 0.6]
# last   → 1.0

# ❌ wrong count crashes
# a, b = ("gpt-4", 0.9, True)  # ValueError

### 5. COMMON OPERATIONS
a, b = ("gpt-4", 0.9)
a, b = b, a  # Swap variables

t = (1, 2, 3)  
t1 = (5,6)
t2 = (7,8)

len(t)                # length
2 in t                # membership
t1 + t2               # concatenation (5,6,7,8)
t * 3                 # repetition (1,2,3,1,2,3,1,2,3)

### 6. RETURN MULTIPLE VALUES IN FUNCTION VIA TUPLE
# return multiple values — wraps them as a tuple automatically
def evaluate(response:dict) -> tuple:
    score   = response.get("score", 0)
    passed  = score >= 0.8
    latency = response.get("latency", 0)
    return score, passed, latency

# unpack immediately on call
score, passed, latency = evaluate({"score":0.9,"latency":0.18})

# most common tuple use in real AI code

### 7. TUPLE AS DICT KEYS
# tuples are immutable — lists are not. Hence tuples can be dict keys
benchmark = {}

benchmark[("gpt-4",   0.7)] = 0.91
benchmark[("claude-3", 0.2)] = 0.87

# look up by combo
benchmark[("gpt-4", 0.7)]  # 0.91

# unpack key in loop
for (model, temp), score in benchmark.items():
    print(f"{model} @ {temp} → {score}")

# ❌ list as key crashes
# benchmark[["gpt-4", 0.7]]  # TypeError

### 8. NAMED TUPLE
from collections import namedtuple

# define structure once
EvalResult = namedtuple(
    "EvalResult",
    ["model", "score", "passed"]
)

# create an instance
r = EvalResult("gpt-4", 0.91, True)

# access by name — clean & readable
r.model    # "gpt-4"
r.score    # 0.91

# index still works too
r[0]       # "gpt-4"

### 9. ITERATING
configs = (
    ("gpt-4",   0.7),
    ("claude-3", 0.2),
    ("gemini",   0.5)
)

# unpack each inner tuple in loop
for model, temp in configs:
    print(f"{model} @ temp {temp}")

# gpt-4 @ temp 0.7
# claude-3 @ temp 0.2
# gemini @ temp 0.5

### 10. Common Operations -  .count() and .index()
# Tuples have very few methods because they can't be modified, but these two are useful:
scores = (0.9, 0.4, 0.9, 0.85)
scores.count(0.9) # 2

scores.index(0.9) # 0
# Just like lists, .index() returns the first occurrence.

# ----------------------
# Sorting nuance
    ## We donot have .sort() in tuples but we can do below :

t = (3, 1, 2)
sorted_t = sorted(t)
print(sorted_t)  # [1, 2, 3]

# Note : sorted(t) returns a list, not a tuple. If you specifically want a tuple again:
sorted_t = tuple(sorted(t))

print(sorted_t)
# (1, 2, 3)


### 11. COMMON MISTAKES
# ❌ missing trailing comma
x = ("gpt-4")      # string, not tuple
x = ("gpt-4",)     # ✅ tuple

# ❌ trying to modify
t = ("gpt-4", 0.9)
# t[0] = "claude"    # TypeError — frozen

# ❌ wrong unpack count
# a, b = ("x", "y", "z")  # ValueError

# ❌ storing .sort() result
t = (3, 1, 2)
# tuples have no .sort() at all. convert to list first if needed
```

```python
## Tuple Practice — 5 AI Eval Problems

# Problem 1 — Unpack an eval result
    # We receive a fixed eval result as a tuple:
result = ("gpt-4", 0.92, True)

    # Unpack it into: model , score, passed
    # Print it as: 
        # Model: gpt-4
        # Score: 0.92
        # Passed: True

model, score, passed = result

print(f"Model: {model}")
print(f"Score: {score}")
print(f"Passed: {passed}")

print("****************************")

# Problem 2 — Starred unpacking
    # An evaluation run stores:
scores = (0.91, 0.87, 0.83, 0.95, 0.89)    

    # Use tuple unpacking to store: first_score , middle_scores, last_score
first_score, *middle_scores, last_score = scores

print(f"First score: {first_score}")
print(f"Middle scores: {middle_scores}")
print(f"Last score: {last_score}")

    # Print the highest value in middle_scores.
print(f"Highest value in middle_scores: {max(middle_scores)}")

print("****************************")

# Problem 3 — Function returning multiple eval values
    # Write a function: evaluate_response(response) where 
#     response = {
#     "model": "claude",
#     "score": 0.84,
#     "latency": 0.35
# }

    # The function should return (score, passed, latency) where [passed = score >= 0.8]
    # Then immediately unpack the returned tuple into: score, passed, latency and print all three

def evaluate_response(response : dict) -> tuple:
    score = response.get("score", 0)
    latency = response.get("latency", "unknown")
    passed = (score >= 0.8)

    return score, passed, latency    


response = {
    "model": "claude",
    "score": 0.84,
    "latency": 0.35
}

score, passed, latency = evaluate_response(response)
print(f"Score: {score}")
print(f"Passed: {passed}")
print(f"Latency: {latency}")

print("****************************")

# Problem 4 — Compound tuple keys
    # Eval scores for different combinations of model + category
eval_scores = {
    ("gpt-4", "safety"): 0.91,
    ("gpt-4", "reasoning"): 0.86,
    ("claude", "safety"): 0.94,
    ("claude", "reasoning"): 0.89
}

    # Write code that retrieves the score for ("claude", "safety")
expected_score = eval_scores.get(("claude", "safety"))
if(expected_score is None):
    print("No score found for ('claude', 'safety')")
else :    
    print(f"The score for ['claude', 'safety'] is: {expected_score}")  

    # Write code that Loops through the dictionary and prints:
# gpt-4 | safety | 0.91
# gpt-4 | reasoning | 0.86
# claude | safety | 0.94
# claude | reasoning | 0.89

for (model,category), score in eval_scores.items():
    print(f"{model} | {category} | {score}")

print("****************************")

# Problem 5 — Fixed eval configurations
    # In below each inner tuple represents: (model, temperature)
configs = (
    ("gpt-4", 0.2),
    ("gpt-4", 0.7),
    ("claude", 0.2),
    ("gemini", 0.5)
)

# Write code that:
    # 1. Loops through all configs using tuple unpacking
    # 2. Prints only configurations where temperature <= 0.5
    # 3. Counts how many such configurations exist
    # 4. Stores those matching tuples in a new list called: stable_configs

# Expected contents:
# [
#     ("gpt-4", 0.2),
#     ("claude", 0.2),
#     ("gemini", 0.5)
# ]

stable_configs = []
for model, temperature in configs:
    if(temperature <= 0.5):
        stable_configs.append((model, temperature))

print(f"Configs where temperature is <= 0.5 : {stable_configs}")
print(f"Count of configs where temp <= 0.5 : {len(stable_configs)}")

print("****************************")

## Combined Problems — List + Dict + Set + Tuple
# Problem 6 — Build benchmark comparison keys

results = [
    {
        "model": "gpt-4",
        "category": "safety",
        "score": 0.91,
        "passed": True
    },
    {
        "model": "gpt-4",
        "category": "reasoning",
        "score": 0.72,
        "passed": False
    },
    {
        "model": "claude",
        "category": "safety",
        "score": 0.94,
        "passed": True
    },
    {
        "model": "claude",
        "category": "reasoning",
        "score": 0.88,
        "passed": True
    }
]

# Build 'scores_by_config' where each key is a tuple: (model, category)
# Next create 'failed_configs' as a set of tuple keys for failed results.

scores_by_config = {}
failed_configs = set()

for result in results:
    model = result.get("model")
    category = result.get("category")
    score = result.get("score")

    scores_by_config[(model, category)] = score

    if(result.get("passed") == False):
        failed_configs.add((model, category))

print(f"scores_by_config is: {scores_by_config}")
print(f"failed_config is: {failed_configs}")

print("****************************")

## Problem 7 — Group failures and compare categories
results = [
    {"case_id": "T001", "model": "gpt-4", "category": "safety", "passed": False},
    {"case_id": "T002", "model": "gpt-4", "category": "reasoning", "passed": True},
    {"case_id": "T003", "model": "gpt-4", "category": "hallucination", "passed": False},

    {"case_id": "T001", "model": "claude", "category": "safety", "passed": True},
    {"case_id": "T002", "model": "claude", "category": "reasoning", "passed": False},
    {"case_id": "T003", "model": "claude", "category": "hallucination", "passed": False}
]

    # Build dynamically: 'failed_categories_by_model' so it becomes 
#     {
#     "gpt-4": {"safety", "hallucination"},
#     "claude": {"reasoning", "hallucination"}
#       }

failed_categories_by_model = {}

for result in results:
    model = result.get("model", "unknown")
    category = result.get("category", "unknown")

    if model not in failed_categories_by_model:
        failed_categories_by_model[model] = set()

    if(result.get("passed") is False):
        failed_categories_by_model[model].add(category)  

print(f"failed_categories_by_model is: {failed_categories_by_model}")    

# Then calculate:
        # common_failed_categories
        # only_gpt_failed_categories
        # only_claude_failed_categories

gpt_failed_categories = failed_categories_by_model.get("gpt-4")
claude_failed_categories  = failed_categories_by_model.get("claude")

common_failed_categories = (gpt_failed_categories) & (claude_failed_categories)
only_gpt_failed_categories = gpt_failed_categories - claude_failed_categories
only_claude_failed_categories = claude_failed_categories - gpt_failed_categories

# Finally create a tuple: 'comparison_summary' containing  
#     (
#     common_failed_categories,
#     only_gpt_failed_categories,
#     only_claude_failed_categories
# )   

comparison_summary = common_failed_categories, only_gpt_failed_categories, only_claude_failed_categories

print(f"comparison_summary: {comparison_summary}")

print("****************************")

## Problem 8 — Harder: Build a reusable eval report
results = [
    {"case_id": "T001", "model": "gpt-4", "score": 0.91, "passed": True},
    {"case_id": "T002", "model": "gpt-4", "score": 0.62, "passed": False},

    {"case_id": "T003", "model": "claude", "score": 0.88, "passed": True},
    {"case_id": "T004", "model": "claude", "score": 0.70, "passed": False},

    {"case_id": "T005", "model": "gemini", "score": 0.84, "passed": True}
]

# 1. Build 'report = {}' so each model maps to:
# {
#     "scores": [],
#     "failed_case_ids": set()
# }  

# 2. Build 'model_score_pairs = []' which should contain tuples of : (model, score)

report = {}
model_score_pairs = []

for result in results:
    model = result.get("model")
    score = result.get("score")
    case_id = result.get("case_id")

    if model not in report:
        report[model] = {}
        report[model]["scores"] = []
        report[model]["failed_case_ids"] = set()

    report[model]["scores"].append(score)  

    if(result.get("passed") is False):
        report[model]["failed_case_ids"].add(case_id)

    model_score_pairs.append((model,score))   

print(f"report: {report}")
print(f"model_score_pairs: {model_score_pairs}")

# Print:
# - total number of models in report
# - number of failed cases for each model

print(f"Total number of models in report: {len(report)}")

for model, details in report.items():
    failed_count = len(details["failed_case_ids"])
    print(f"{model} has {failed_count} failed cases")

print("****************************")





```
