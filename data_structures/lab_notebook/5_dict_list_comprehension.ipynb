Quick Notes :
### List comprehension
1. [expr for x in items]                    # basic
2. [expr for x in items if condition]       # with filter
3. [a if condition else b for x in items]   # with transform

### Dict comprehension
4. {k: v for k, v in d.items()}             # basic
5. {k: v for k, v in d.items() if condition} # with filter
6. {k: expr for k, v in d.items()}          # transform values

Note - zip() pairs up two lists item by item and is used commonly. See Q4 below for example

### Nested Comprehensions (Advanced)
matrix = [[1,2],[3,4]]
7. flatten = [num for row in matrix for num in row]


```python
### Practise questions on above 7 concepts

## List comprehension basic 
# Q1.1 : You have a list of LLM API responses. Extract just the model names into a new list.
responses = [
    {"model": "gpt-4",    "score": 0.9},
    {"model": "claude-3", "score": 0.4},
    {"model": "gemini",   "score": 0.85},
]

model_names = [r["model"] for r in responses]
print(model_names) # Expected output: ["gpt-4", "claude-3", "gemini"]

# Q1.2 : You have token lengths. Create a list of token lengths using list comprehension.
tokens = ["AI", "model", "test"]
token_length = [len(t) for t in tokens]
print(f"Token length list:  {token_length}") # Token length list:  [2, 5, 4]

# ---------------------------------------------------------------------------
## List comprehension with filter 
# Q2.1 : From the same responses list, extract only the models that scored 0.8 or above.
filtered_model = [r["model"] for r in responses if r["score"] >= 0.8]
print(filtered_model) # Expected output: ["gpt-4", "gemini"]

# Q2.2 : You have prediction confidences. Extract only confidences greater than 0.8
confs = [0.91, 0.45, 0.82, 0.30]
high_confs = [c for c in confs if c >= 0.8]
print(f"Confidence greater than 0.8 list:  {high_confs}") # [0.91, 0.82]

# ---------------------------------------------------------------------------
## List comprehension with transform
# Q3 : You have a list of scores. Label every score as "pass" if >= 0.8, else "fail". Every item must appear in the output — nothing filtered out.
pythonscores = [0.9, 0.4, 0.85, 0.6, 1.0]

score_label = ["pass" if score >=0.8 else "fail" for score in pythonscores]
print(score_label) # Expected output: ["pass", "fail", "pass", "fail", "pass"]

# ---------------------------------------------------------------------------
## Dict comprehension basic
# Q4.1 : You have two separate lists — case IDs and their scores. Build a single dict that maps each case ID to its score.
case_ids = ["case_1", "case_2", "case_3"]
scores   = [0.9, 0.4, 0.85]

id_score = {id : score for id, score in zip(case_ids, scores)}
print(id_score) # o/p - {'case_1': 0.9, 'case_2': 0.4, 'case_3': 0.85}

# Q4.2 : You have tokens. Create a dictionary: 
# "ai" → 2
# "ml" → 2
# "dl" → 2
# (length of each token)
tokens = ["ai", "ml", "dl"]
tokens_dict = {t:len(t) for t in tokens}
print(f"Token dict with length: {tokens_dict}") # {'ai': 2, 'ml': 2, 'dl': 2}

# ---------------------------------------------------------------------------
## Dict comprehension with filter
# Q5 : You have a dict of eval scores. Return a new dict containing only the cases that failed (score below 0.8).
scores = {
    "case_1": 0.9,
    "case_2": 0.4,
    "case_3": 0.85,
    "case_4": 0.6,
    "case_5": 1.0,
}

failed_cases = {case : score for case, score in scores.items() if score < 0.8}
print(failed_cases) # o/p - {'case_2': 0.4, 'case_4': 0.6}

# ---------------------------------------------------------------------------
## Dict comprehension transform values
# Q6 : You have a dict of raw scores between 0 and 1. Transform every value into a percentage string rounded to no decimal places.
pythonscores = {"case_1": 0.9, "case_2": 0.4, "case_3": 0.85}

new_scores = {c : f"{id * 100:.0f}%" for c, id in pythonscores.items()}
print(new_scores) # o/p - {'case_1': '90%', 'case_2': '40%', 'case_3': '85%'}

# ---------------------------------------------------------------------------
## Nested comprehension
# Q7 : You ran 3 eval batches. Each batch is a list of scores. Flatten all scores into one single list.
eval_batches = [
    [0.9, 0.4, 0.85],
    [0.6, 1.0],
    [0.75, 0.88, 0.5],
]

batch = [score for batch in eval_batches for score in batch]
print(batch) # o/p - [0.9, 0.4, 0.85, 0.6, 1.0, 0.75, 0.88, 0.5]

```
