### What is a Virtual Environment? 
- A virtual environment is an isolated Python installation for your project. It has its own packages, its own dependencies, completely separate from every other project on your machine.

### PART 1 — Creating a Virtual Environment
Command : ``python -m venv venv``

This creates a venv/ folder in your project containing sub-folders like : 
    bin/               ← Python executable + pip
    lib/               ← installed packages go here
    pyvenv.cfg         ← config file

### PART 2 — Activating the Virtual Environment
Command : ``source venv/bin/activate``

After activation your terminal prompt changes :
#### Before activation
supritis@MacBook Python %

#### After activation
(venv) supritis@MacBook Python %

#### Deactivating  
- Command : ``deactivate``
- Prompt goes back to normal — you are back to global Python.

### PART 3 — Installing Packages
- ``pip`` is Python's package manager. Stands for - 'pip install packages'
- Syntax : ``pip install ...``

Examples :
1. Install a single package : ``pip install pandas``
2. Install specific version - ``pip install pandas==2.0.0``
3. Install multiple packages at once : ``pip install pandas numpy pytest``
4. Upgrade existing package : ``pip install --upgrade pandas``
5. Uninstall : ``pip uninstall pandas``

Important Note : Always activate ``venv`` BEFORE ``pip install ..``
- Wrong order : installs globally, not in venv
pip install pandas
source venv/bin/activate

- Correct order : installs into isolated venv
source venv/bin/activate
pip install pandas

#### Checking installed packages
- ``pip list`` : List all installed packages in active venv
- ``pip show pandas`` : Check specific package version and details

### PART 4 — requirements.txt
- ``requirements.txt`` is a plain text file that lists all packages your project needs with their versions. 
- It is the standard way to share dependencies with your team or deploy to a server.

#### Creating requirements.txt file
- Command : In terminal under your venv type - ``pip freeze > requirements.txt``
- `` > `` saves that output to ``requirements.txt`` 

#### Installing from requirements.txt
- Your team can run one command to install everything on their system : ``pip install -r requirements.txt``

### The standard project setup flow
- Step 1 : create project folder
``mkdir my_eval_project``
``cd my_eval_project``

- Step 2 : create virtual environment
``python -m venv venv``

- Step 3 : activate it
``source venv/bin/activate``

- Step 4 : install packages you need
``pip install pandas numpy pytest requests python-dotenv anthropic``

- Step 5 : save dependencies
``pip freeze > requirements.txt``

- Step 6 : start coding

Note - The ``venv/`` folder should never be committed to git. Add it to .gitignore
