# Terminal commands for Folders & file operations

    1. pwd : print (current) working directory

    2. ls : list contents of current directory (excluding the hidden files)
    2.1 ls -a : list contents of current directory including hidden files

    3. cd : change directory

    4. mkdir  : make directory (creates a new folder)

    5. touch : creates an empty file. If file already exists, cmd updates it's timestamp without deleting contents.

    6. mv : move or rename a file. Useful for organzing files.
        Example: mv test_model.py test_llm.py  # renames file to test_llm.py
        Example: mv draft_report.txt ../reports # synatx : mv <source> <destination>

    > Imp note: In a terminal path . means the current directory and .. is the parent directory. This is imp when navigating paths.

    7. cp - stands for copy. This copies a file. Original file remains. Cmd useful for creating backups.
        Syntax : cp <source_file> <destination_file> 

    8. cat - Read a file. Example: cat eval_cases.jsonl (displays all content in the file)       

    9. echo with > - Write text into a file. Overwrites past content.
        Example: echo '{"prompt":"What is 2+2?","expected":"4"}'  > eval_cases.jsonl

    10. echo with >> - Write text into a file. Appends text along with exisitng content
        Example: echo '{"prompt":"What is 4+4?","expected":"8"}'  >> eval_cases.jsonl

    11. nano <filename> - Create and Edit Text Files. Nano is a text editor that runs directly
    inside the terminal. It can create a new file(if doesnot already exist), open an existing file, 
    add-delete-modify text interactively.

    12. wc -l : 'wc' stands for word count, and '-l' tells it to count lines.
        Example : wc -l eval_cases.jsonl  # O/P :  3 eval_cases.jsonl (means 3 jsonl lines)

 ## Difference Between touch, echo, cat, and nano
  - touch : Create empty file or update timestamp
  - echo with > : Write text, replacing previous content   
  - echo with >> : Append text to existing content
  - cat : Read and display file contents
  - nano : Interactively create, read, and edit files

# Absolute vs. Relative Paths
There are 2 ways to access a file :
    1. Absolute Path - The complete address, starting from the filesystem root /.
        Example : cat /Users/supritis/Documents/AI_Learnings/terminal_practise/datasets/eval_cases.jsonl
    2. Relative Path - An address interpreted relative to your current working directory.
        Example : cat eval_cases.jsonl (assuming you are in ../terminal_practise/datasets/ already)

    You can also write : cat ~/Documents/AI_Learnings/terminal_practise/datasets/eval_cases.jsonl 
    as ~ represents your home directory - /Users/supritis.

    How to know your home directory ? -  echo $HOME

# Running Python Scripts from Terminal
- Our goal is to run Python programs directly from our Mac Terminal, without clicking the Run button in VS Code.

- Basic syntax from Terminal : python3 filename.py
    - 'python3' invokes the Python interpreter.
    - 'filename.py' specifies the script to execute.
    - The terminal displays whatever the script prints, along with any errors.

Example :
> supritis@Sana-MacBook-Pro terminal_practise % pwd
> /Users/supritis/Documents/AI_Learnings/Python/terminal_practise
> supritis@Sana-MacBook-Pro terminal_practise % echo 'print("EvalKit practise runner started")' > run_eval.py
> supritis@Sana-MacBook-Pro terminal_practise % python3 run_eval.py 
> EvalKit practise runner started

# Important Python CLI Commands
- ``python3 script.py`` : Execute a Python script   
- ``python3 -c 'print(10+20)'`` : Execute inline Python code
- ``python3 --version`` : Check Python version
- ``which python3`` : Locate the Python executable selected by the shell
- ``python3 -m pip list`` : List installed packages
