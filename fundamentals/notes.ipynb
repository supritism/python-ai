```python
# Topic 1 : Variables

## Syntax : variable_name = value
## A variable is a name tag bound to an object in memory. 
## Python is dynamically typed meaning we donot have to declare the type in advance unlike in Java

score = 80
print(type(score)) # o/p - <class 'int'>

## AI evaluation variables examples
prompt = "What is the capital of France?"
expected_answer = "Paris"
model_answer = "Paris"
score = 1
passed =True

# Topic 2 : Basic Data types

## There are 4 basic data types - int, float, boolean, str
    # int - whole numbers like 10, -5, 500
    # float - decimal numbers 0.7, 98.5
    # str - text like "hello"
    # bool - True or False

## There is also a Special value called 'None'. 
    # None is Python's way of saying - “This variable exists, but it currently points to no meaningful value.”
    # It is not the same as 0, False or ""

model_response = None
print(type(model_response)) # o/p - <class 'NoneType'>

if model_response is None:
    print("No response recieved")
   

# Topic 3: Strings - extremely important in AI

## AI applications work with enormous amounts of text, so strings matter.
model = "Claude"
message = "Evaluating model: " + model

print(message) # Evaluating model: Claude

## In modern Python, we usually use f-strings
model = "GPT-4"
score = 95

print(f"{model} has a score of {score}") # GPT-4 has a score of 95

## Python also gives Strings useful build-in operations. Example :
model_response = " Paris "
model_response.lower().strip()
print(model_response) # Paris 

result = model_response.lower().strip() 
print(result) # paris

## IMPORTANT - Strings in Python are immutable. So methods like .lower() or .strip() donot modify the orginal String.
## They return a new String. So if the new value is not stored anywhere, Python discards it.

# Topic 4: Operators

## 4.1 Arithmetic Operator : + , - , * , / , % , **
a = 10
b = 3
print(f" a + b = {a+b}")
print(f" a - b = {a-b}")
print(f" a * b = {a*b}")
print(f" a/b = {a/b}") # This gives the quotient
print(f" a % b = {a%b}") # This gives the reminder
print(f" a ** b = {a**b}") # This is 'to the power of'

## 4.2 Assignment Operator : = , += , -=, *=
c = 8
c += 1 # c = c + 1
c -= 1 # c = c - 1
c *= 1 # c = c * 1
c /= 2 # c = c/2

## 4.3 Comparison Operator : 
    #  ==, != , > , < , >= , <=
    # Returns either 'True' or 'False'

model_op_accuracy = 0.85
expected_accuracy = 0.80

passed = model_op_accuracy >= expected_accuracy
print(passed) # True

## 4.4 Logical Operator : and , or , not
    # 'and' : Both conditions must be true
    #  'or' : Atleast 1 condition must be true
    #  'not' : Reverse a boolean


# Topic 5: Conditionals : Making decisions using 'if' statements

    ## Basic Structure :
    #   if condition:
    #       do_something()

    ## Three types of Conditionals : if , if-else, elif

score = 78

if score >= 90:
    result = "Exceeds the bar"
elif score >= 70:
    result = "Meets the bar"    
else:
    result = "Below the bar"    

print (f"Score {result}")

# Topic 6: Truthiness

## Python sometimes treats values themselves as True or False like below : 
    # "" - False
    # "Hello" - True
    # 0 - False
    # 1 - True
    # None - False

model_response = None

if model_response:
    print("Model returned something")
else:
    print("Model response was empty")    


# Topic 7: 'for' Loops

## Syntax : 
    # for item in sequence:
    #     code to repeat

for number in [1,3,6,9]:
    print(number)   

for i in range(5):
    print(i) # 0, 1, 2, 3, 4

for i in range(1,6): # range(start, stop)
    print(i) # 1, 2, 3, 4, 5    


# Topic 8: Counting inside a loop

passed = 0
if i in range(5):
    passed += 1

print(passed) # 5

# Topic 9: while loops

## 'for' loop - repeat this for a known number of times or across some collection.
## 'while' loop - keep repeating while given condition is true.

attempt = 1

while (attempt <= 3):
    print(f"Attempt: {attempt}")
    attempt += 1

# Topic 10: 'break' and 'continue'

## 'break' immediately stops the loop.
for i in range(10):
    if (i == 5):
        break

    print(i)
# o/p - 0, 1, 2, 3, 4         

## 'continue' skips the current iteration and moves to the next one
for i in range(5):
    if (i == 2):
        continue

    print(i)
# o/p - 0, 1, 3, 4

#----------------------------------------------------------------------
```
